# VS Code Notebook Shortkeys (Jupyter)


## Modes (most important)
- **Esc**: Command mode (cell-level shortcuts active)
- **Enter**: Edit mode (typing inside cell)


## Cell Type / Language
- **M** (command mode): Change selected cell to **Markdown**
- **Y** (command mode): Change selected cell to **Code**
- **Ctrl+Shift+P** -> **Notebook: Change Cell Language**: set code cell language (e.g., **R**, Python, Julia)


## Run Cells
- **Shift+Enter**: Run cell and move to next
- **Ctrl+Enter**: Run selected cell and stay
- **Alt+Enter**: Run cell and insert a new one below


## Insert / Delete / Move
- **A** (command mode): Insert cell **above**
- **B** (command mode): Insert cell **below**
- **D, D** (command mode): Delete selected cell
- **Alt+Up / Alt+Down**: Move selected cell up/down


## Selection / Clipboard
- **C** (command mode): Copy cell
- **X** (command mode): Cut cell
- **V** (command mode): Paste cell below
- **Z** (command mode): Undo last cell operation

## Rounding
- **round( ... ,3)** rounds the response to 3 decimal places

## Printing a string
- **cat("Name: ", ... , "\n", sep = "")** to see Name: ...(e.g result of function)
