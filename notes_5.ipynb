## 5.1 Cross Validation



Device for using the same training data to tell you how well your prediction method works

### Test error
avg error that results from using a statistical learning method to predict the response on a new observation (data not used for training)
### Training error
calculated by applying the statistial learning method to the observations used in its training

Note! Training error can dramatically underestimate the test error

## Validation Set Approach
-dividing the full training data set in two at random, training on one half and testing by predicting the 2nd half
-usually gives a good idea of where the optimal degree of polynomial, but not a good indication of test error as its so variable depending on where the data was split
-can be a little wasteful with only using half the data to train, plus results in a higher error rate because it has been trained with less data

### k-Fold Cross Validation

Divide the data into k groups of approximately equal size (usually between 5 and 10)

e.g if we had 5 parts, we train on 4 and predict the 5th and record the error, then repeat for each group. We then take an average of the prediction errors

$$CV_{(k)} = \frac{1}{k}\sum_{i=1}^{k} MSE_i$$
